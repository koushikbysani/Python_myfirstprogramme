```python
# Common Conventions :

#Variable Naming**: Use lowercase with underscores (`_`) to separate words (e.g., `my_variable`, `student_name`). This is known as `snake_case`.
#Function Naming**: Similar to variables, use lowercase with underscores (e.g., `calculate_average`, `get_student_details`).
#Class Naming**: Use `CamelCase` (first letter of each word capitalized) for class names (e.g., `StudentData`, `GradeCalculator`).
#Constant Naming**: Use all uppercase with underscores for constants (e.g., `MAX_STUDENTS`, `PI`).
#Indentation**: Use 4 spaces for each level of indentation. Avoid tabs.
#Docstrings**: Use triple quotes (`"""Docstring goes here."""`) to write documentation for modules, functions, classes, and methods.
#Comments**: Use `#` for single-line comments. Keep comments concise and relevant.
#Blank Lines**: Use blank lines to separate logical sections of your code for readability.
#Import Statements**: Place import statements at the top of the file, one import per line, and group them (standard library, third-party, local modules).
#Indendation matter a lot
```

```python
|###| Type | Representation | Example |
###||-----|---|---|
###|| `NoneType` | `None` | `value = None` |
###|| `bool` | `True`, `False` | `passed = True` |
###|| `int` | whole number | `age = 25` |
###|| `float` | decimal number | `price = 99.5` |
###|| `complex` | number with `j` | `number = 2 + 3j` |
###|| `str` | `' '`, `" "`, `''' '''`, `""" """` | `name = "Koushik"` |
###|| `list` | `[ ]` | `marks = [80, 75, 90]` |
###|| `tuple` | `( )` or commas | `point = (10, 20)` |
###|| `set` | `{ }` | `subjects = {"Maths", "Science"}` |
###|| `frozenset` | `frozenset()` | `fixed = frozenset({1, 2})` |
###|| `dict` | `{key: value}` | `student = {"name": "Koushik", "age": 25}` |
###|| `range` | `range()` | `numbers = range(1, 6)` |
###|| `bytes` | `b" "` | `data = b"hello"` |
###|| `bytearray` | `bytearray()` | `data = bytearray(b"hello")` |
###|| `memoryview` | `memoryview()` | `view = memoryview(b"hello")` |
```

```python

# Python Basics
#
# Variables
# * Used to store information/data
# * Variables can hold different types of values.
# * No need to declare the data type explicitly.
#
# Data Types
# * int - Whole numbers
# * float - Decimal numbers
# * str - Text/string
# * bool - True or False
# * list - Ordered + changeable collection
# * tuple - Ordered + fixed collection
# * set - Collection of unique elements
# * dict - Key-value pair collection
#
# Strings
# * Used to store text.
# * Brackets: Single quotes '' or double quotes ""
# * Ordered and immutable.
# * Functions learned:
#     * len()
#     * upper()
#     * lower()
#     * strip()
#     * replace()
#     * split()
#
# Lists
# * Ordered + changeable collection
# * Very flexible - you can make changes to the collection.
# * Brackets: Square brackets []
# * Functions learned:
#     * insert()
#     * remove()
#     * append()
#     * pop()
#     * len()
#
# Tuples
# * Ordered + fixed collection
# * Cannot be changed after creation.
# * Brackets: Round brackets ()
# * Immutable - this is a property of tuples.
# * Functions learned:
#     * len()
#     * count()
#     * index()
#
# Sets
# * Collection of unique elements
# * Duplicate elements are not allowed.
# * Brackets: Curly brackets {}
# * Unordered collection.
# * Functions learned:
#     * add()
#     * remove()
#     * discard()
#     * pop()
#     * union()
#     * intersection()
#
# Dictionaries
# * Stores information in key-value pairs.
# * Brackets: Curly brackets {}
# * Each key should be unique.
# * Values can be of different data types.
# * Dictionaries are changeable.
# * Functions learned:
#     * keys()
#     * values()
#     * items()
#     * get()
#     * update()
#     * pop()
#
# Operators
# * Used to perform operations on values.
# * Arithmetic operators:
#     * +
#     * -
#     * *
#     * /
#     * %
#     * //
#     * **
# * Comparison operators:
#     * ==
#     * !=
#     * >
#     * <
#     * >=
#     * <=
# * Logical operators:
#     * and
#     * or
#     * not
#
# Input & Output
# * print() - Displays information on the screen.
# * input() - Takes information from the user.
#
# Type Conversion
# * Used to convert one data type into another.
# * Functions:
#     * int()
#     * float()
#     * str()
#     * bool()
#     * list()
#     * tuple()
#     * set()
#
# Conditional Statements
# * Used to make decisions based on conditions.
# * Keywords:
#     * if
#     * elif
#     * else
#
# Loops
# * Used to repeat a block of code.
# * for loop - Used to iterate through a collection.
# * while loop - Repeats while a condition is true.
# * Keywords:
#     * for
#     * while
#     * break
#     * continue
#     * pass
#
# Functions
# * Reusable block of code.
# * Helps avoid repeating the same code.
# * Defined using def.
# * Can accept parameters/arguments.
# * Can return a value using return.
#
# Nested Data Structures
# * Different data structures can be stored inside each other.
# * A list can contain dictionaries.
# * A dictionary can contain lists.
# * A list can contain tuples, sets, or other lists.
#
# Comments
# * Used to explain code.
# * Python ignores comments while executing the program.
# * Single-line comment: #
#
# Import
# * Used to bring functionality from modules into our program.
# * Keyword: import
#
# Exception Handling
# * Used to handle errors without stopping the entire program.
# * Keywords:
#     * try
#     * except
#     * finally
#     * raise
#
# File Handling
# * Used to read from and write to files.
# * Functions:
#     * open()
#     * read()
#     * write()
#     * close()
# * Common modes:
#     * r - Read
#     * w - Write
#     * a - Append
```
