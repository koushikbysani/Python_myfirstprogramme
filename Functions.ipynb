### Comprehensive Notes on Python Functions

#### 1. What is a Function?
A function is a reusable block of code that performs a specific task. Functions help in modularizing code, making it more organized, readable, and reusable.

#### 2. Defining a Function
You define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`.

```python
def greet_user():
    """This is a docstring describing the function."""
    print("Hello, Welcome to the App!")
```

#### 3. Parameters and Arguments
*   **Parameters**: Variables listed in the function definition.
*   **Arguments**: Values sent to the function when it is called.

```python
def add_numbers(a, b): # a and b are parameters
    print(a + b)

add_numbers(10, 5) # 10 and 5 are arguments
```

#### 4. The `return` Statement
To let a function send a value back to the caller, use the `return` statement. If no return is specified, the function returns `None`.

```python
def multiply(x, y):
    return x * y

result = multiply(4, 3)
```

#### 5. Types of Arguments
*   **Positional Arguments**: Arguments that need to be in the correct order.
*   **Keyword Arguments**: Arguments passed by name, allowing any order.
*   **Default Parameters**: Assigning a default value if no argument is provided.

```python
def power(base, exponent=2):
    return base ** exponent

print(power(3))    # Uses default: 9
print(power(3, 3)) # Overwrites default: 27
```

#### 6. Variable Scope
*   **Local Scope**: Variables created inside a function belong only to that function.
*   **Global Scope**: Variables created in the main body of the script are accessible everywhere.

#### 7. Why use Functions?
1. **DRY (Don't Repeat Yourself)**: Write once, use many times.
2. **Abstraction**: Hide the complexity and focus on the result.
3. **Testability**: Easier to test small parts of code individually.

````python
def calculate_gst(price,gst_percentage):
  gst = price * gst_percentage /100
  return gst
````

````python
def calculate_delivery_charges(price,delivery_charges):
  delivery_charges = price * delivery_charges /100
  return delivery_charges
````
