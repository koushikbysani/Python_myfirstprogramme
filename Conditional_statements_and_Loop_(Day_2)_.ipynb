## Summary of Python Topics Covered

Based on the notebook cells, here is a detailed breakdown of the concepts explored:

### 1. Variables and Data Types
* **Assignment:** Storing values like integers (`age = 34`, `pin = 3455`) and strings (`day = "Wednesday"`).
* **User Input:** Using `input()` to capture data from the user and `int()` to convert that input into a number.

### 2. Conditional Statements (Control Flow)
* **If-Else:** Used for binary decisions (e.g., checking if a PIN is correct or if a user is old enough to drive).
* **Elif (Else-If):** Used for multiple branching conditions, such as assigning grades based on marks.
* **Logical Operators:**
    * `and`: Requires both conditions to be true (e.g., age >= 25 AND having a license).
    * `or`: Requires at least one condition to be true (e.g., Saturday OR Sunday).
* **Comparison Operators:** Using `==` (equal to), `>=` (greater than or equal to), and `<` (less than) to compare values.

### 3. Loops (Iteration)
* **For Loops:** Used to iterate over a sequence (like a `range`). Examples include printing numbers from 1 to 100 or processing a specific list of customers.
* **Range Function:**
    * `range(stop)`: Starts from 0.
    * `range(start, stop)`: Starts and ends at specific numbers.
    * `range(start, stop, step)`: Increments by a specific 'step' value (e.g., printing only even or odd numbers).
* **While Loops:** Repeats a block of code as long as a condition remains true (e.g., counting until 10).

### 4. Loop Control Statements
* **Break:** Immediately exits the loop when a condition is met (e.g., stopping the loop when `i` reaches 6).
* **Continue:** Skips the current iteration and moves to the next one (e.g., skipping the number 6 while printing 1 to 10).

### 5. String Manipulation
* **`.lower()`:** Used to make user input case-insensitive (e.g., treating 'Even', 'EVEN', and 'even' the same way).

```python
# TOPIC 1: VARIABLES AND DATA TYPES

# Terminology: 'Variable' is an 'Identifier' (it identifies a value in memory).
# Convention: Python uses 'snake_case' (all lower case with underscores).

# Choosing 'product_name' over 'pn' makes the code 'Self-Documenting' (easy to read).
product_name = "Mechanical Keyboard"  # String (str): Textual data.

# Choosing 'float' for 'price' allows for decimal precision needed in currency.
price = 89.99                        # Float: Decimal numbers.

# Choosing 'int' for 'stock_quantity' prevents 'Invalid Logic' (you can't have half a keyboard).
stock_quantity = 50                  # Integer (int): Whole numbers.

# Convention: Boolean names start with 'is', 'has', or 'can' to act as 'Flag Indicators'.
is_in_stock = True                   # Boolean (bool): True/False state.
```

```python
# TOPIC 2: CONDITIONAL LOGIC (Control Flow)

# Convention: 'age' is a 'Descriptive Variable'. 'has_student_id' is a 'Boolean Flag'.
age = 22
has_student_id = True

# Terminology: 'if/elif/else' are 'Branching Statements'.
if age < 5:
    price = 0
elif age < 18 or has_student_id:
    # Synonym: 'or' is a 'Logical Operator'. It creates a 'Compound Condition'.
    price = 10
else:
    # Terminology: The 'Default Case' or 'Fallback'.
    price = 20

print(f"Your ticket price is ${price}")
```

```python
# TOPIC 3: LOOPS (For & While)

# Synonym: 'playlist' is a 'List'. In other languages, this is an 'Array' or 'Collection'.
# Convention: Use 'Plural Identifiers' (songs, items, playlist) for lists.
playlist = ["Song A", "Song B", "Song C", "Song D"]

# 'For' Loop: Also called 'Definite Iteration' because we know the count.
# Convention: The iterator 'song' is the 'Singular' version of the list name 'playlist'.
print("Playing Playlist:")
for song in playlist:
    print(f"Now playing: {song}")

# 'While' Loop: Also called 'Indefinite Iteration' (loops until a condition changes).
# Terminology: 'battery_level' is the 'Loop Control Variable'.
battery_level = 95
while battery_level < 100:
    print(f"Charging... {battery_level}%")
    # Terminology: 'Incrementing' the variable. Without this, you get an 'Infinite Loop'.
    battery_level += 1
print("Fully Charged!")
```

```python
# TOPIC 4: BREAK AND CONTINUE

# Terminology: 'items' is the 'Data Structure' (a List of Strings).
items = ["Box 1", "Target", "Box 2", "Empty", "Box 3"]

for item in items:
    # Terminology: 'Condition Check'.
    if item == "Empty":
        # Terminology: 'Skip' or 'Next Iteration'.
        continue  # Skips the rest of the code inside the loop for THIS specific item.

    print(f"Checking {item}...")

    if item == "Target":
        # Terminology: 'Early Exit' or 'Terminating the Loop'.
        print("Found it! Stopping search.")
        break  # Stops the loop completely for 'Computational Efficiency'.
```

# Python Naming Conventions & Terminology Guide

To code like a professional, it is important to understand not just the syntax, but the language used to describe it and the rules for naming your components.

### 1. Understanding the 'Playlist' (Data Structures)
In Topic 3, we used `playlist = ["Song A", "Song B"]`.

**Synonyms & Terms:**
*   **List:** The official Python term for this structure. It is an ordered collection that can be changed.
*   **Array:** In other languages (like C or Java), this is called an Array. You will often hear programmers use these terms interchangeably.
*   **Collection / Sequence:** General terms for any variable that holds multiple pieces of data.
*   **Element / Item:** Each individual entry inside the list (e.g., "Song A" is an *element*).
*   **Index:** The position of an item (starting at 0). "Song A" is at index 0.

### 2. Naming Conventions (Identifiers)
Variables, like `product_name` or `has_student_id`, follow specific 'Naming Conventions'.

**The Rules:**
*   **Snake Case:** In Python, we use `snake_case` (all lowercase, words joined by underscores).
    *   *Correct:* `user_choice`, `stock_quantity`
    *   *Incorrect:* `UserChoice` (PascalCase) or `userChoice` (camelCase).
*   **Booleans (The 'Is/Has' Rule):** When naming true/false variables, start with a verb to make it an 'indicator'.
    *   *Examples:* `is_in_stock`, `has_licence`, `is_valid`.
*   **Pluralization:** Use plural names for collections.
    *   *Examples:* `items`, `songs`, `customers`.

### 3. Variable Naming Logic
Why did we choose specific names in the examples?

*   **`ip` vs `user_choice`:** In cell `D1QDgh_lrJW3`, `ip` was used as a shortcut for 'Input'. However, `user_choice` is a better variable name because it is **descriptive**.
*   **`i` vs `customer`:** In simple loops, we use `i` (short for 'index' or 'iterator'). But when the loop represents something specific, we use the singular version of the collection: `for customer in customers:`.

# 🚀 Section 3: Progressive Practice (10 Examples)
These examples progress from simple memory assignment to complex logic flow.

```python
# Example 1: Basic Variable Assignment
# Logic Breakdown:
# 1. The computer sets aside a memory slot labeled 'user_name'.
# 2. It places the text "Alex" inside that slot.
# 3. The f-string {} looks into that slot, 'pulls out' the value, and injects it into the print sentence.
user_name = "Alex"
print(f"Hello, {user_name}!")
```

```python
# Example 2: Integer vs Float Calculation
# Logic Breakdown:
# 1. We multiply an 'int' (5) by a 'float' (19.95).
# 2. Python Logic: To avoid losing decimal data, Python 'promotes' the whole number to a decimal internally.
# 3. Calculation: 5.0 * 19.95 = 99.75.
# 4. The result is stored as a float because it contains a fractional part.
items_ordered = 5
price_per_item = 19.95
total_cost = items_ordered * price_per_item
print(f"Total: ${total_cost}")
```

```python
# Example 3: Boolean Logic / Inequality
# Logic Breakdown:
# 1. input() gets text, int() turns it into a math-ready number.
# 2. Calculation: Python asks: "Is the value in 'age' 18 or higher?"
# 3. If yes, it creates the value 'True'. If no, it creates 'False'.
# 4. This binary answer is stored in the variable 'is_adult'.
age = int(input("Enter your age: "))
is_adult = age >= 18
print(f"Is the user an adult? {is_adult}")
```

```python
# Example 4: The 'If-Else' Decision Gate
# Logic Breakdown:
# 1. The 'if' statement acts like a gatekeeper. It evaluates the math: (25 > 30).
# 2. Result: False.
# 3. Because it's False, Python 'jumps over' the first print and enters the 'else' block instead.
temperature = 25
if temperature > 30:
    print("It's a hot day.")
else:
    print("It's a pleasant day.")
```

```python
# Example 5: Multi-way 'elif' Logic
# Logic Breakdown:
# 1. Check 1: (85 >= 90) -> False. Move to next.
# 2. Check 2: (85 >= 70) -> True.
# 3. Execution: Once a 'True' is found, Python runs that block and IGNORES the rest of the 'else' logic entirely.
score = 85
if score >= 90:
    print("Excellent")
elif score >= 70:
    print("Good")
else:
    print("Needs Improvement")
```

```python
# Example 6: Compound Logic ('or')
# Logic Breakdown:
# 1. Python looks at the first part: (65 >= 60) -> True.
# 2. Because this is an 'OR' statement, Python knows the result is True as soon as it sees one True.
# 3. It doesn't even need to check 'is_member'—the whole condition becomes True immediately.
is_member = True
age = 65
if age >= 60 or is_member:
    print("Discount Applied")
else:
    print("Standard Pricing")
```

```python
# Example 7: The 'For' Loop Mechanism
# Logic Breakdown:
# 1. Python looks at the List (the collection).
# 2. Round 1: 'fruit' becomes "Apple". Print.
# 3. Round 2: 'fruit' becomes "Banana". Print.
# 4. The loop automatically stops when it hits the end of the collection.
fruits = ["Apple", "Banana", "Cherry"]
for fruit in fruits:
    print(f"I like {fruit}")
```

```python
# Example 8: Step-based Range Calculation
# Logic Breakdown:
# 1. start=2, end=11, step=2.
# 2. Calculation: Start at 2. Add the step (2). New value = 4. Add the step (2). New value = 6.
# 3. This continues until the next calculation would be >= 11.
print("Even numbers sequence:")
for num in range(2, 11, 2):
    print(num)
```

```python
# Example 9: While Loop / Counter Logic
# Logic Breakdown:
# 1. Python checks: (0 < 3)? Yes -> Run code.
# 2. Inside: attempts becomes 0 + 1 = 1.
# 3. Back to top: (1 < 3)? Yes -> Run code. attempts becomes 2.
# 4. Back to top: (2 < 3)? Yes -> Run code. attempts becomes 3.
# 5. Final check: (3 < 3)? NO. Exit loop.
attempts = 0
while attempts < 3:
    print("Connecting to server...")
    attempts += 1
print("Connection established.")
```

```python
# Example 10: Nested Logic and 'Break' Efficiency
# Logic Breakdown:
# 1. Python starts iterating 1, 2, 3...
# 2. In each cycle, it performs a comparison: (i == 7).
# 3. On cycle 7: The 'if' becomes True.
# 4. 'break' is triggered. Python instantly 'kills' the loop and jumps to the very end of the code block, ignoring numbers 8 through 19.
target_value = 7
for i in range(1, 20):
    if i == target_value:
        print(f"Target {i} found! Stopping loop.")
        break
    else:
        print(f"Searching... checked {i}")
```

# 🍔 Section 4: Real-World Case Study (Food Delivery App)
Applying all concepts into a single logical workflow for a delivery service.

```python
delivery_address = "123 Main St"  # Logic: Define a 'Descriptive Identifier' for a String. We use 'snake_case' to keep it readable for other developers.
print(f"Delivery set to: {delivery_address}")  # Logic: Print an eff-string: Delivery set to, followed by the value of delivery address injected.
```

```python
food_price = 12.50  # Logic: Store as a 'Float' because prices have decimals; using an 'int' here would lose the cents.
delivery_fee = 2     # Logic: Store as an 'Integer' because this specific fee is a whole number.
total_bill = food_price + delivery_fee  # Logic: 'Arithmetic Operation'. Python turns the 2 into 2.0 (Implicit Casting) so the math stays precise.
print(f"Amount to pay: ${total_bill}")  # Logic: Print an eff-string: Amount to pay, followed by the value of total bill.
```

```python
min_order_threshold = 15.00  # Logic: We define a 'Threshold' variable so we can easily change the minimum order limit later in one place.
is_minimum_met = total_bill >= min_order_threshold  # Logic: Use a 'Comparison Operator' (>=) to get a Boolean (True/False) state indicator.
print(f"Can proceed to checkout? {is_minimum_met}")  # Logic: Print an eff-string with the value of is minimum met injected.
```

```python
distance_km = 7  # Logic: Define distance as an Integer for simple kilometer tracking.
if distance_km > 5:  # Logic: A 'Branching Statement'. If the Boolean expression is True, the program enters this specific 'Code Block'.
    total_bill += 3.00  # Logic: 'Augmented Assignment'. This is shorthand for updating the existing value in memory by adding 3.0.
    print("Long-distance surcharge applied.")
else:  # Logic: The 'Fallback Block'. This only runs if the 'if' condition fails (is False).
    print("Standard logistics fee applied.")
```

```python
promo_code = "SAVE10"  # Logic: A String identifier representing user input.
if promo_code == "SAVE20":  # Logic: Compare two Strings. If they aren't identical, Python skips to the next branch.
    total_bill *= 0.80
elif promo_code == "SAVE10":  # Logic: An 'Else-If' check. Python only checks this if the first 'if' was False.
    total_bill *= 0.90  # Logic: Multiply the total by 0.9 to apply a 10% discount.
else:  # Logic: The 'Default' branch if no promotion codes match.
    print("Invalid promotion identifier.")
```

```python
is_premium_user = True  # Logic: A 'Boolean Flag'. Using 'is_' tells other coders this is a True/False status.
if is_premium_user or total_bill > 30:  # Logic: 'Logical OR Operator'. If the user is premium, Python doesn't even bother checking the bill amount.
    print("Logistics status: Free Delivery Applied!") # Logic: This runs because the 'Compound Condition' resulted in True.
```

```python
order_items = ["Burger", "Fries", "Soda"]  # Logic: A 'List' (or Array). We use a plural name because it holds a 'Collection' of data.
print("Receipt Details:")
for item in order_items:  # Logic: 'Definite Iteration'. 'item' is a temporary variable that holds one element at a time during the loop.
    print(f"- {item}")  # Logic: Print an eff-string with the value of item injected.
```

```python
print("Estimated dispatch windows:")
for minutes in range(10, 40, 10):  # Logic: The 'range' function generates a 'Sequence' (10, 20, 30). It stops BEFORE the final number (40).
    print(f"Arriving in approx {minutes} mins...")  # Logic: Print an eff-string with the value of minutes injected.
```

```python
is_driver_assigned = False  # Logic: Initialize the 'Loop Control Variable' to a False state.
search_attempts = 0  # Logic: A 'Counter' variable to track how many times the loop has run.
while not is_driver_assigned:  # Logic: 'Indefinite Iteration'. This keeps running as long as the status is NOT True.
    search_attempts += 1  # Logic: Increment the counter by 1 in each cycle.
    print(f"Scanning for courier... attempt {search_attempts}")
    if search_attempts == 3:  # Logic: 'Exit Condition'. We simulate finding a driver after 3 tries.
        is_driver_assigned = True  # Logic: Re-assign the variable to True, which tells the 'while' loop to stop.
print("Courier successfully assigned!")
```

```python
courier_ids = [101, 505, 707, 808]  # Logic: A 'Collection' of Integer identifiers stored in a List.
for c_id in courier_ids:  # Logic: Iterate through the IDs one by one.
    if c_id == 505:  # Logic: Check for a specific 'Element Match'.
        print(f"Courier {c_id} is in range. Engaging...")  # Logic: Print an eff-string with the value of courier id injected.
        break  # Logic: 'Early Exit'. We stop the loop immediately to save computer memory once we find the right courier.
    print(f"Scanning courier ID {c_id}...")
```
