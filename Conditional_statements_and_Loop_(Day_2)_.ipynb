```python
#Ex 2: I user Input Even then print Even numbers , if odd is entered then print Odd numbers

# Combining both If and for Loop

user_choice = input("Enter 'even' or 'odd': ")

if user_choice.lower() == "even":
  print("Printing even numbers:")
  for num in range(0, 21, 2):
    print(num)
elif user_choice.lower() == "odd":
  print("Printing odd numbers:")
  for num in range(1, 20, 2):
    print(num)
else:
  print("Invalid input. Please enter 'even' or 'odd'.")
```
